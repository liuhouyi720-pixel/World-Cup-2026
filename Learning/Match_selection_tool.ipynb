Now we find all the matches in La Liga in 2018/2019 season. Let's first take a look at the constructure of its correspondent match.json.

```python

from pathlib import Path
DATA_DIR = Path("D:/UIUC/Projects/World Cup 2026/archive/data")
match_file = DATA_DIR / "matches"/str(COMPETITION_ID) / f"{SEASON_ID}.json"
df_match = pd.read_json(match_file)
df_match.head()
```

We noticed that there are many dictionary under the column like the home_team and away_team. I want to normalize this dataframe and make it looks better.

```python
from pandas import json_normalize
# Split the "home_team" dictionary into multiple columns and concatenate them with the original df.
df_home = df_match['home_team'].apply(pd.Series)
df_home = df_home.rename(columns={
    "home_team_id":"home_team_id",
    "home_team_name":"home_team_name"
})
# away_team同理
df_away = df_match['away_team'].apply(pd.Series)
df_away = df_away.rename(columns={
    "away_team_id":"away_team_id",
    "away_team_name":"away_team_name"
})
# 合并回原表，可按需删除原来的字典列
df_match = pd.concat([df_match, df_home, df_away], axis=1)
# 如果不需要原始字典列，可执行：df.drop(columns=["home_team","away_team"],inplace=True)
```

First, we do a seasonal analysis. These data could be very easy to get if you open any football website. Like what's the ranking of the team and what is the total goals of this team in this single season

```python
def get_barca_res(row):
    if row["home_team_name"] =="Barcelona":
        barcelona_score = row["home_score"]
        opponent_score = row["away_score"]
    else:
        barcelona_score = row["away_score"]
        opponent_score = row["home_score"]  
    if barcelona_score > opponent_score:
        return "Win"    
    if barcelona_score < opponent_score:
        return "Lose"
    if barcelona_score == opponent_score:
        return "Draw"

barca_matches["barca_result"] = barca_matches.apply(get_barca_res, axis=1)
 #add a new column summarize the result of Barcelona

result_counts = barca_matches["barca_result"].value_counts()
win_count = result_counts.get("Win", 0)
lose_count = result_counts.get("Lose", 0)   
draw_count = result_counts.get("Draw", 0)

#calculate the total credit points for Barcelona based on the results
total_points = win_count * 3 + draw_count * 1 + lose_count * 0

print(f"Barcelona's total points in {index_season} season {index_league}: {total_points}")
```

Now we can count the total goals and the total conceded goals of a team in one season. And the average data.

```python
def goals(row):
    if row["home_team_name"] =="Barcelona":
        barcelona_score = row["home_score"]
        opponent_score = row["away_score"]
    else:
        barcelona_score = row["away_score"]
        opponent_score = row["home_score"]  
    return barcelona_score, opponent_score

barca_matches[["barca_goals","opponent_goals"]] = barca_matches.apply(goals, axis=1, result_type="expand")

total_goals_scored = barca_matches["barca_goals"].sum()
total_goals_conceded = barca_matches["opponent_goals"].sum()
print(f"Barcelona's total goals scored in {index_season} season {index_league}: {total_goals_scored}")
print(f"Barcelona's total goals conceded in {index_season} season {index_league}: {total_goals_conceded}")

#The average goals scored and conceded per match
average_goals_scored = total_goals_scored / len(barca_matches)
average_goals_conceded = total_goals_conceded / len(barca_matches)
print(f"Barcelona's average goals scored per match in {index_season} season {index_league}: {average_goals_scored:.2f}")
print(f"Barcelona's average goals conceded per match in {index_season} season {index_league}: {average_goals_conceded:.2f}")
```

```python
def split_match(row):
    #home team records:
    home_goals = row["home_score"]
    away_goals = row["away_score"]
    if home_goals > away_goals:
        home_pts = 3
        away_pts = 0
    elif home_goals == away_goals:#the second condition should be used elif, or the the first condition is ignored.
        home_pts = 1
        away_pts = 1
    else:
        home_pts = 0
        away_pts = 3

    home_row = {'team': row["home_team_name"], 'goals_scored': home_goals, 'goals_conceded': away_goals, 'points': home_pts}
    away_row = {'team': row["away_team_name"], 'goals_scored': away_goals, 'goals_conceded': home_goals, 'points': away_pts}
    return [home_row, away_row]

match_list = []#create an empty list to store the match data for each team
for index, row in df_match.iterrows():
    #iterrows() is used to iterate over the rows of the DataFrame, returning both the index and the row data as a Series.
    match_list.extend(split_match(row))
team_df = pd.DataFrame(match_list)

team_df.head()
```

```python
league_table = team_df.groupby("team", as_index=False).agg(
    total_scored = ('goals_scored', 'sum'),
    total_conceded = ('goals_conceded', 'sum'), 
    total_points = ('points', 'sum')
)
league_table = league_table.sort_values(by='total_points', ascending=False)
league_table['goals_difference'] = league_table['total_scored']-league_table['total_conceded']
league_table['average_goals_scored'] = league_table['total_scored']/38
league_table.head(20)
```
